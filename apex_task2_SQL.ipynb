## joining_both tables

```python
query = """
SELECT
    c."Customer Name",
    c.Segment,
    o."Order ID",
    o.Sales,
    o.Category
FROM customers c
JOIN orders o
ON c."Customer ID" = o."Customer ID";
"""

pd.read_sql_query(query, conn).head(10)
```

advanced_sql

sub query

```python
query = """
SELECT
    "Order ID",
    "Product Name",
    Sales
FROM sales
WHERE Sales > (
    SELECT AVG(Sales)
    FROM sales
)
ORDER BY Sales DESC;
"""

pd.read_sql_query(query, conn).head(10)
```

##CTE_WITH

```python
query = """
WITH product_sales AS (
    SELECT
        "Product Name",
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY "Product Name"
)

SELECT *
FROM product_sales
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

##RANK

```python
query = """
SELECT
    "Product Name",
    Sales,
    RANK() OVER (
        ORDER BY Sales DESC
    ) AS Sales_Rank
FROM sales
LIMIT 20;
"""

pd.read_sql_query(query, conn)
```

##ROW

```python
query = """
SELECT
    "Product Name",
    Sales,
    ROW_NUMBER() OVER (
        ORDER BY Sales DESC
    ) AS Row_Number
FROM sales
LIMIT 20;
"""

pd.read_sql_query(query, conn)
```

##LAG

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS Month,
        SUM(Sales) AS Total_Sales
    FROM sales
    GROUP BY Month
)

SELECT
    Month,
    Total_Sales,
    LAG(Total_Sales) OVER (
        ORDER BY Month
    ) AS Previous_Month_Sales
FROM monthly_sales
ORDER BY Month;
"""

pd.read_sql_query(query, conn)
```

##VIEW

```python
query = """
CREATE VIEW IF NOT EXISTS high_value_sales AS
SELECT
    "Order ID",
    "Customer Name",
    "Product Name",
    Sales
FROM sales
WHERE Sales > 1000;
"""

conn.execute(query)
conn.commit()

print("View created successfully!")
```

```python
query = """
SELECT *
FROM high_value_sales
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
#1 top 5 products by sales




query2 = """
SELECT *
FROM (
    SELECT 
        "Product ID",
        Category,
        Sales,
        RANK() OVER (
            PARTITION BY Category 
            ORDER BY Sales DESC
        ) AS top_product
    FROM df
) t
WHERE top_product <= 5
ORDER BY Category, top_product
"""

res = duckdb.sql(query2).df()

res
```

```python
#2 Top 5 customers by total sales


query = """
SELECT 
    "Customer Name",
    SUM(Sales) AS total_sales
FROM df
GROUP BY "Customer Name"
ORDER BY total_sales DESC
LIMIT 5
"""

res = duckdb.sql(query).df()

res
```

```python
#3 TOP 5 PRODUCT BY SALES


query = """ 
select
"Product Name",
sum(sales) as total_sales 
from df 
group by "Product Name"
order by total_sales desc
limit 5
"""

res = duckdb.sql(query).df()
res
```

```python
#4 SALES BY REGION


query = """ 
select Region,
sum(sales) as Total_sales 
from df 
group by Region
order by Total_sales DESC
limit 5

"""

res = duckdb.sql(query).df()
res
```

```python
#5 sales by category 


query = """ 
select Category,
sum(sales) as total_sales 
from df 
group by Category 
order by total_sales desc
limit 5

"""
res = duckdb.sql(query)
res
```

```python
#7 top 5 sub-categories 


query = """
SELECT 
    "Sub-Category",
    SUM(Sales) AS total_sales
FROM df 
GROUP BY "Sub-Category"
ORDER BY total_sales DESC
LIMIT 5
"""

res = duckdb.sql(query).df()
res
```

```python
#top 5 product in each category 
query8 = """
SELECT *
FROM (
    SELECT 
        Category,
        "Product Name",
        SUM(Sales) AS total_sales,
        RANK() OVER (
            PARTITION BY Category
            ORDER BY SUM(Sales) DESC
        ) AS product_rank
    FROM df
    GROUP BY Category, "Product Name"
) t
WHERE product_rank <= 5
ORDER BY Category, product_rank
"""

res8 = duckdb.sql(query8).df()

res8
```
